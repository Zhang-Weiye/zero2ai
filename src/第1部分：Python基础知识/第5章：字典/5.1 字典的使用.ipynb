```python
# 字典是由一个{}组成，其中包含多个键-值对
# 键即索引的名称，值即索引的内容
aliens = {'color':'green', 'points':5}
print(aliens['color'])
print(aliens['points'])
```

```python
# 访问字典中的值 把键放在字典索引[]中
aliens ={'colors':'green'}
print(aliens['colors'])
```

```python
# 也可将字典的值赋给新的变量
aliens = {'colors':'green', 'points':5}
new_points = aliens['points']
print("You just earned " + str(new_points) + " points!")
```

```python
# 添加键值对
aliens = {'color':'green', 'points':5}
print(aliens)

aliens['x_position'] = 0
aliens['y_position'] = 25
print(aliens)
```

```python
# 向空字典添加键值对来创建字典
aliens = {}
aliens['color'] = 'green'
aliens['points'] = 5
print(aliens)
```

```python
# 修改字典中的值
aliens = {'color':'green'}
print("The alien is " + aliens['color'] + ".")

aliens['color'] = 'yellow'
print("The alien is now " + aliens['color'] + ".")
```

```python
# 删除键值对 del
alien = {'color':'green', 'points':5}
print(alien)

del alien['color']
print(alien)
```

```python
# 由类似对象组成的字典
favorite_languages = {
    'jen': 'python',
    'sarah': 'c',
    'edward': 'ruby',
    'phil': 'python',
}

print("Sarah's favorite language is " +
      favorite_languages['sarah'].title() + ".")
```

```python
# 习题6-1
person = {
    'family name': 'King',
    'given name': 'Domdom',
    'age': 4,
    'city': 'Nanyang'
}
for key,value in person.items():
    print(key, ":", value)
```

```python
# 习题6-3
alphabet = {
    'if': '条件',
    'for': '循环',
    'import': '导入',
    'in': '存在',
    'return': '返回', 
}
for key, value in alphabet.items():
    print(key, ":", value)
```
