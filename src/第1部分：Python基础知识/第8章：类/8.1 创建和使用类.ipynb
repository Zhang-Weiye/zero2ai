```python
# 创建一个小狗类
class Dog():
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def sit(self):
        print(self.name.title() + " is now sitting.")
        
    def roll_over(self):
        print(self.name.title() + " rolled over!")
```

```python
# 根据类创建实例
my_dog = Dog('willie', 6)

print("My dog's name is " + my_dog.name.title() + ".")
print("My dog is " + str(my_dog.age) + " years old.")
```

```python
# 访问类的属性
age = my_dog.age
name = my_dog.name
print(f"my dog's name is {name}, age is {age}")
```

```python
# 调用类的方法
my_dog.sit()
my_dog.roll_over()
```

```python
# 创建多个实例
my_dog = Dog('willie', 6)
your_dog = Dog('lucy', 3)

print("My dog's name is " + my_dog.name.title() + ".")
print("My dog's age is " + str(my_dog.age) + ".")
my_dog.sit()

print("Your dog's name is " + your_dog.name.title() + ".")
print("Your dog's age is " + str(your_dog.age) + ".")
your_dog.sit()
```

```python
# 习题9-1
class Restaurant():
    def __init__(self, restaurant_name, cuisine_type):
        self.restaurant_name = restaurant_name
        self.cuisine_type = cuisine_type
        
    def describe_restaurant(self):
        print(f"My restaurant's name is {self.restaurant_name}.")
        print(f"Our cuisine type is {self.cuisine_type}.")
        
    def open_restaurant(self):
        print("We're open now.")
        
my_restaurant = Restaurant('xiao huajiao', 'fried chicken')
my_restaurant.describe_restaurant()
my_restaurant.open_restaurant()
```

```python
# 习题9-3
class User():
    def __init__(self, first_name, last_name, **user_info):
        self.first_name = first_name
        self.last_name = last_name
        for key, value in user_info.items():
            self.key = value
    
    def describe_user(self):
        print(f"first name: {self.first_name} , last name: {self.last_name}")
        for k, v in self.__dict__.items():
            if k not in ['first_name', 'last_name']:
                print(f"key: {k} , value: {v}")
    
    def greet(self):
        print(f"Hello, {self.first_name.title()} {self.last_name.title()} !")
        
user1 = User('bruce', 'lee', age=18, gender='man')
user2 = User('jacky', 'chan', occupation='actor')

user1.describe_user()
user1.greet()

user2.describe_user()
user2.greet()

```
