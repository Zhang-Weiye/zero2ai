```python
# 利用sort()方法进行永久排序，按首字母顺序
cars = ['bwm', 'audi', 'toyota', 'sabaru']
print(cars)

cars.sort()
print(cars)
```

```python
# 若要按首字母相反顺序排序，需要向sort()传入参数 reverse = True
print(cars)
cars.sort(reverse=True)
print(cars)
```

```python
# 使用函数sorted()对列表进行排序
# 排序后是不影响原始的数据的
cars = ["bwm", "audi", "sabaru", "toyota"]

print(f"Here is the original list:\n{cars}")
print(f"Here is the sorted list:\n{sorted(cars)}")
print(f"Here is the original list again:\n{cars}")
```

```python
# 将原始列表倒序排列 reverse()

cars = ['audi', 'bwm', 'toyota']
print(cars)

cars.reverse()
print(cars)
```

```python
# 确定列表的长度
cars = ['audi', 'bwm', 'toyota']
num = len(cars)
print(num)
```
