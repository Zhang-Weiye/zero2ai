```python
# 元组是由()括住其中的元素来定义的
dimensions = (200, 50)
print(dimensions[0])
print(dimensions[1])
```

```python
# 元组禁止修改其中的值
dimensions = (200, 50)
dimensions[0] = 250
```

```python
# 遍历元组中所有的值
dimensions = (1, 2, 3, 4)
for dimension in dimensions:
    print(dimension)
```

```python
# 虽然不能修改元组的元素，但是可以通过重新赋值的方式修改整个元组
dimensions = (200, 50)
print("Original dimensions:")
for dimension in dimensions:
    print(dimension)
    
dimensions = (400, 50)
print("Modified dimensions:")
for dimension in dimensions:
    print(dimension)
```
