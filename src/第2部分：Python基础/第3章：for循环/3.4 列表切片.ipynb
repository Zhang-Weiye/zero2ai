```python
# 从起始位置0开始，到索引3之前的切片
players = ['charles', 'martina', 'michael', 'florence', 'eli']
print(players[0:3])
```

```python
# 从起始位置1开始，到索引4之前的切片
players = ['charles', 'martina', 'michael', 'florence', 'eli']
print(players[1:4])
```

```python
# 如果没有指定第一个索引，将自动从起始开始
players = ['charles', 'martina', 'michael', 'florence', 'eli']
print(players[:4])
```

```python
# 如果没有指定末尾索引，将自动到终止位置
players = ['charles', 'martina', 'michael', 'florence', 'eli']
print(players[1:])
```

```python
# 遍历切片
players = ['charles', 'martina', 'michael', 'florence', 'eli']

print("Here are the first three players on my team:")
for player in players[:3]:
    print(player.title())
```

```python
# 复制列表
my_foods = ['pizza', 'falafel', 'carrot cake']
friend_foods = my_foods[:]

print(f"My favorite foods are: {my_foods}")
print(f"My friend's favorite foods are: {friend_foods}")
```

```python
# 对不同列表添加不同元素，用切片
my_foods = ['pizza', 'falafel', 'carrot cake']
friend_foods = my_foods[:]

my_foods.append('ice cream')
friend_foods.append('cannoli')


print(f"My favorite foods are: {my_foods}")
print(f"My friend's favorite foods are: {friend_foods}")
```

```python
# 如果不用切片赋值,则两个变量等同，各自添加元素时都会添加上
my_foods = ['pizza', 'falafel', 'carrot cake']
friend_foods = my_foods

my_foods.append('ice cream')
friend_foods.append('cannoli')


print(f"My favorite foods are: {my_foods}")
print(f"My friend's favorite foods are: {friend_foods}")
```

```python
# 习题4-10
companies = ['ali', 'deepseek', 'bytedance', 'baidu', 'tecent', 'xiaomi', 'kuaishou']
print(f"The first three items in the list are: {companies[:3]}")
print(f"Three items from the middle of the list are: {companies[int(len(companies)/2)-1:int(len(companies)/2)+2]}")
print(f"The last three items in the list are: {companies[-3:]}")
```

```python
# 习题4-11
my_foods = ['pizza', 'falafel', 'carrot cake']
friend_foods = my_foods

my_foods.append('ice cream')
friend_foods.append('cannoli')


print(f"My favorite foods are: {my_foods} \n")
print("My favorite foods are:")
for food in my_foods:
    print(food, end=",")

print(f"My friend's favorite foods are: {friend_foods} \n")
print("My friend's favorite foods are:")
for food in friend_foods:
    print(food, end=",")
```
