```python
# 简单的示例
cars = ['audi', 'bmw', 'subaru', 'toyota']

for car in cars:
    if car == 'bmw':
        print(car.upper())
    else:
        print(car.title())
```

```python
# 检查是否相等
car = 'bmw'
print(car == 'bmw')

car = 'audi'
print(car == 'bwm')
```

```python
# 检查是否相等时需要考虑大小写
car = 'Audi'
print(car == 'audi')
print(car.lower() == 'audi')
```

```python
# 检查是否不相等
requested_topping = 'mushrooms'

if requested_topping != 'anchovies':
    print("Hold the anchovies!")
```

```python
# 比较数字
age = 18
print(age == 18)
```

```python
# 检查数字是否不相等
answer = 17
if answer != 42:
    print("That is not the correct answer. Please try again!")
```

```python
# 条件语句中可以包含各种数学比较符号
age = 19
print(age<21)
print(age<=21)
print(age>21)
print(age>=21)
```
