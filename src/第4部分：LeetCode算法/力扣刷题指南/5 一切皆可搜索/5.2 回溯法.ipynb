```python
"""
46 Permutations
给定一个无重复数字的整数数组，求其所有的排列方式
"""
# 输入是一个一维整数数组，输出是一个二维数组

def backtracking(nums: List[int], level: int, permutations: List[List[int]]):
    if level == len(nums) - 1:
        permutations.append(nums[:])
        return 
    for i in range(level, len(nums)):
        nums[i], nums[level] = nums[level], nums[i]
        backtracking(nums, level + 1, permutations)
        nums[i], nums[level] = nums[level], nums[i]

def permute(nums: List[int]) -> List[List[int]]:
    permutations = []
    backtracking(nums, 0, permutations)
    return permutations
```

```python
"""
理解提示：

递归 = 树的一层
for循环 = 当前层的所有分支

append → 递归 → pop
就是：
选择 → 进入下一层 → 回溯
"""


def backtracking(
        combinations: List[List[int]],  # 存储最终所有组合
        pick: List[int],                # 当前已经选择的数字（当前路径）
        pos: int,                       # 下一次可选择数字的起始位置
        n: int,
        k: int
):
    """
    回溯函数：构造所有组合

    参数说明：
    pick : 当前路径（已经选择的数字）
    pos  : 下一次可选择的起始数字
    n    : 最大数字
    k    : 需要选择的数字数量

    例如：
    pick = [1]
    pos = 2

    表示：
    下一层只能从 2,3,4...n 中选
    """

    # 终止条件：已经选够 k 个数字
    if len(pick) == k:
        # 注意要复制，否则后面 pop 会修改结果
        combinations.append(pick[:])
        return

    # 在当前层尝试所有可能的数字
    for i in range(pos, n + 1):

        # 1️⃣ 做选择
        pick.append(i)

        # 2️⃣ 进入下一层搜索
        # 下一层只能选择比 i 更大的数字
        backtracking(combinations, pick, i + 1, n, k)

        # 3️⃣ 撤销选择（回溯）
        pick.pop()


def combine(n: int, k: int) -> List[List[int]]:
    """
    主函数：返回所有组合
    """
    combinations = []  # 存放所有组合
    pick = []          # 当前路径

    # 从数字 1 开始搜索
    backtracking(combinations, pick, 1, n, k)

    return combinations
```
