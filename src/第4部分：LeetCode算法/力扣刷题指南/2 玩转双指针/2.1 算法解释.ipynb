```python
"""
167 Two Sum 2 - Input array is sorted
在一个增序的整数数组里找到两个数，使它们的和为给定值。已知有且只有一对解。
"""

# 输入是一个数组（numbers）和一个给定值（target) 输出是两个数的位置，从1开始计数

def twoSum(numbers: List[int], target: int) -> List[int]:
    # 初始化双指针，一个在最右端，一个在最左端
    l, r = 0, len(numbers) - 1
    # 当左指针小于右指针
    while l < r:
        # 求和当前两个指针所指位置
        two_sum = numbers[l] + numbers[r]
        # 如果恰好等于目标值，则直接退出
        if two_sum == target:
            break
        # 如果小于目标值，则左指针加1
        elif two_sum < target:
            l += 1
        # 如果大于目标值，则右指针减1
        else:
            r -= 1
    # 数组从0开始下标，因此返回具体位置时要加1
    return [l+1, r+1]

numbers = [2, 7, 11, 15]
twoSum(numbers, 9)
```

```python
"""
88 Merge Sorted Array
给定两个有序数组，把两个数组合并为一个
"""

# 输入是两个数组和它们分别的长度m 和n。其中第一个数组的长度被延长至m +n，多出的 n 位被0 填补。题目要求把第二个数组归并到第一个数组上

def merge(nums1: List[int], m: int, nums2: List[int], n: int) -> None:
    # pos-每次要插入的位置指针，m和n是指向两个数组最大（最后）的指针
    pos = m + n - 1
    m -= 1
    n -= 1
    # 在把其中一个数组遍历完之前
    while m >= 0 and n >= 0:
        # 把两个数组中最大的数字添加到nums1的末尾
        if nums1[m] > nums2[n]:
            nums1[pos] = nums1[m]
            m -= 1
        else:
            nums1[pos] = nums2[n]
            n -= 1
        # 位置指针向前移动一个位置
        pos -= 1
    # 如果nums1被复制完，则把nums2指针n之前的所有数字复制到nums1,切片是左闭右开
    nums1[: n + 1] = nums2[: n + 1]
    # 如果nums2被复制完，则不需要再操作nums1(因为已经被排好序)

nums1 = [1, 2, 3, 0, 0, 0]
nums2 = [2, 5, 6]
merge(nums1, 3, nums2, 3)
nums1
```

```python
"""
76 Minimum Window Substring
给定两个字符串 s 和t，求 s 中包含t 所有字符的最短连续子字符串的长度，同时要求时间复杂度不超过O(n)
"""

# 输入是两个字符串s和t， 输出是一个S字符串的子串。如果不存在解，则输出一个空字符串。

def minWindow(s: str, t: str) -> str:
    # 统计t中的字符情况，等价于：
    # freq = dict()
    # for c in t:
    #     freq[c] = freq.get(c, 0) + 1
    freq = Counter(t)
    # 移动滑动窗口，不断更改统计数据
    count = 0
    min_l, min_length = None, None
    l = 0
    for r in range(len(s)):
        if s[r] not in freq:
            continue
        freq[s[r]] -= 1
        if freq[s[r]] >= 0:
            count += 1
        # 滑动窗口已包含t中全部字符，尝试右移l，在不影响结果的情况下寻找最短串
        while count == len(t):
            if min_length == None or r - l + 1 < min_length:
                min_l = l
                min_length = r - l + 1
            # 把l位置的字符移除频率统计，并检查t中对应字符是否重新缺失。
            if s[l] in freq:
                freq[s[l]] += 1
                if freq[s[l]] > 0:
                    count -= 1
            l += 1
    return "" if min_length is None else s[min_l: min_l + min_length]
```

```python
from collections import Counter
# 把freq看做是欠款的账单，每个字符欠多少
def minWindow(s: str, t: str) -> str:
    # 1. 欠账单：正数=还欠多少；0=刚好够用；负数=富余
    freq = Counter(t)
    print("欠账单初始值:", dict(freq))

    count = 0               # 已“刚好够用”的字符种数
    min_l, min_len = None, None
    l = 0

    # 2. 右指针一直往右扩
    for r, ch in enumerate(s):
        print("\n>>> r={} 字符 '{}'".format(r, ch))
        if ch not in freq:
            print("  无关字符，跳过")
            continue

        freq[ch] -= 1
        if freq[ch] == 0:   # 刚好从“欠”变“不欠”
            count += 1
            print("  还款成功，count 升至", count)
        else:
            print("  还款后欠账单:", dict(freq))

        # 3. 一旦全部够用，就尽量缩左指针
        while count == len(t):
            cur_len = r - l + 1
            if min_len is None or cur_len < min_len:
                min_l, min_len = l, cur_len
                print("  ★ 更新最短窗口：左={} 长度={} 子串='{}'"
                      .format(l, cur_len, s[l:r+1]))

            left_ch = s[l]
            if left_ch in freq:
                freq[left_ch] += 1
                if freq[left_ch] == 1:  # 从“不欠”变“欠”
                    count -= 1
                    print("  左指针右移后 '{}' 重新欠账，count 降至".format(left_ch), count)
                else:
                    print("  左指针右移后欠账单:", dict(freq))
            l += 1

    # 4. 返回结果
    return "" if min_len is None else s[min_l:min_l+min_len]


# ------------------ 演示 ------------------
if __name__ == "__main__":
    tests = [
        ("ADOBECODEBANC", "ABC"),
        ("a", "a"),
        ("a", "aa"),
        ("bba", "ab")
    ]
    for s, t in tests:
        print("\n==================== 例题：s='{}'  t='{}' ====================".format(s, t))
        ans = minWindow(s, t)
        print("\n最终答案：'{}'".format(ans))
```
