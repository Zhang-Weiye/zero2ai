```python
"""
69 Sqrt(x)
给定一个非负整数x，求它的开方，向下取整。
"""
# 输入一个整数，输出一个整数
def mySqrt(x: int) -> int:
    # 因为根号x在x > 1 时小于x，所以右边界可以为x
    l, r = 1, x
    mid = (l + r) // 2
    mid_sqr = mid**2
    if mid_sqr == x:
        return mid
    if mid_sqr < x:
        l = mid + 1
    else: 
        r = mid - 1
    return 4

def newtonIter(x: int) -> int:
    # f(t) = t**2 - x = 0
    # t+1 = t - f(t)/f'(t) = t - (t**2 - x)/2t = t - t/2 + x/2t = t/2 + x/2t
    t = x
    while t**2 > x:
        t = (t + x // t) // 2
```

```python
"""
34 Find First and Last Position of Element in Sorted Array
给定一个增序的整数数组和一个值，查找该值第一次和最后一次出现的位置。
"""
# 输入是一个数组和一个值，输出为该值第一次出现的位置和最后一次出现的位置（从0开始）；如果不存在该值，则两个值都设为-1
def lowerBound(nums: List[int], target: int) -> int:
    """_summary_

    Args:
        nums (List[int]): 数组
        target (int): 目标值

    Returns:
        int: 第一个 = target 的数字位置；
             若数组里 没有 target， l 会指向第一个比它大的元素，或 len(nums)
    """
    l, r = 0, len(nums)
    while l < r:
        mid = (l + r) // 2
        if nums[mid] < target:
            l = mid + 1
        else:
            r = mid
    return l

def upperBound(nums: List[int], target: int) -> int:
    """_summary_

    Args:
        nums (List[int]): 数组
        target (int): 目标值

    Returns:
        int: 第一个 > target 的位置
    """
    l, r = 0, len(nums)
    while l < r:
        mid = (l + r) // 2
        if nums[mid] <= target:
            l = mid + 1
        else:
            r = mid
    return l

def searchRange(nums: List[int], target: int) -> List[int]:
    """
    增序数组里，若  nums[i] == target ，则所有  target  必然落在
    [first, last]  一段连续区间，
    左边全  < target ，右边全  > target 。
    于是问题退化成“找这段区间的左右端点”。
    2. 
    端点 ⇔ 最左/最右满足条件的下标
    第一次出现的位置 = 第一个 ≥ target 的下标（lower_bound）
    最后一次出现的位置 = 最后一个 ≤ target 的下标
    而“最后一个 ≤ target” 正是 第一个 > target 的下标再减 1（upper_bound − 1）
    """
    if not nums:
        return [-1, -1]
    # 返回第一个等于target（或者大于target）的位置
    lower = lowerBound(nums, target)
    # 返回最后一个等于target的位置
    upper = upperBound(nums, target) - 1
    # 如果在数组里没有找到target
    if lower == len(nums) or nums[lower] != target:
        return [-1, -1]
    return [lower, upper]
```

```python
"""
162 Find Peak Element
给定一个数组，定义峰值为比所有两边都大的数字，求峰值的位置。
一个数组里可能存在多个峰值，返回任意一个即可。
"""
# 输入是一个数组，输出是峰值的位置
def findPeakElement(nums: List[int]) -> int:
    n = len(nums)
    if n == 1:
        return 0
    if nums[0] > nums[1]:
        return 0
    if nums[n - 1] > nums[n - 2]:
        return n - 1
    l, r = 1, n - 2
    while l <= r:
        mid = (l + r) // 2
        if nums[mid] > nums[mid + 1] and nums[mid] > nums[mid - 1]:
            return mid
        elif nums[mid] > nums[mid - 1]: # 上坡，峰在右
            l = mid + 1
        else: # 下坡，峰在左
            r = mid - 1  
    return -1

# 简洁版
def simpleFindPeak(nums: List[int]) -> int:
    l, r = 0, len(nums) - 1
    while l < r:
        mid = (l + r) // 2
        if nums[mid] > nums[mid+1]: # 下坡，峰在左
            r = mid
        else: # 上坡，峰在右
            l = mid + 1
    return l
```

```python
"""
81 Search in Rotated Sorted Array II
一个原本增序的数组被首尾相连后按某个位置断开
（如[1,2,2,3,4,5]![2,3,4,5,1,2]，在第一位和第二位断开），我们称其为旋转数组。
给定一个值，判断这个值是否存在于这个旋转数组中。
"""
# 输入是一个数组和一个值，输出是一个布尔值，表示数组中是否存在该值。
def search(nums: List[int], target: int) -> bool:
    l, r = 0, len(nums) - 1
    while l <= r:
        mid = (l + r) // 2
        if nums[mid] == target: # 如果中点就是目标值
            return True
        if nums[mid] == nums[l]: # 无法判断哪个区间是增序的，但l一定不是target，不然在上一步就返回了
            l += 1
        elif nums[mid] == nums[r]: # 无法判断哪个区间是降序的，但r一定不是target，不然在上一步就返回了
            r -= 1
        # 现在 nums[l] < nums[mid] 或 nums[mid] < nums[r] 至少成立一个，
        # 可以明确知道哪半边是「严格升序」的，于是用经典二分套路。
        if nums[mid] < nums[r]:    # 右半边有序（nums[mid..r] 升序）
            if nums[mid] < target <= nums[r]:   # target 落在右半边
                l = mid + 1                       # 缩到右半边
            else:                                # target 在左半边
                r = mid - 1                       # 缩到左半边
        else:                       # 左半边有序（nums[l..mid] 升序）
            if nums[l] <= target < nums[mid]:   # target 落在左半边
                r = mid - 1
            else:                                # target 在右半边
                l = mid + 1
    return False
```

```python
"""
81 Search in Rotated Sorted Array II

一个原本增序的数组被首尾相连后按某个位置断开（如[1,2,2,3,4,5]-> [2,3,4,5,1,2]，在第一位和第二位断开），我们称其为旋转数组。
给定一个值，判断这个值是否存在于这个旋转数组中。
"""

# 输入是一个数组和一个值，输出是一个布尔值，表示数组中是否存在该值
def search(nums: List[int], target: int) -> bool:
    left, right = 0, len(nums) - 1
    while left <= right:
        mid = left + (left - right) // 2
        if nums[mid] == target:
            return True
        # 处理特殊情况
        if nums[left] == nums[mid] == nums[right]:
            left += 1
            right -= 1
        # 如果左半部分是有序的
        elif nums[left] < nums[mid]:
            # target落在左半部分
            if nums[left] < target <  nums[mid]:
                right = mid - 1
            # target不在左半部分
            else:
                left = mid + 1
        # 如果右半部分是有序的
        else:
            # target落在右半部分
            if nums[mid] < target < nums[right]:
                left = mid + 1
            # target落在左半部分
            else:
                right = mid - 1
    # 如果什么都没找到，则返回False
    return False
```
