````python
from typing import List
import random
````

````python
"""
快速排序
"""
def quickSort(nums: List[int], l: int, r: int) -> None:
    """
    nums: target list
    l: left position (initialize as 0)
    r: right position (initialize as len(nums) - 1)
    """
    if l >= r:
        return
    pivot = random.randint(l, r)
    pivot_val = nums[pivot]
    nums[l], nums[pivot] = nums[pivot], nums[l]
    i, j = l + 1, r
    while True:
        while i < j and nums[j ] >= pivot_val:
            j -= 1
        while i < j and nums[i] <= pivot_val:
            i += 1
        if i == j:
            break
        nums[i], nums[j] = nums[j], nums[i]
    new_pivot = i if nums[i] <= nums[l] else i - 1
    nums[l], nums[new_pivot] = nums[new_pivot], nums[l]
    quickSort(nums, l, new_pivot - 1)
    quickSort(nums, new_pivot + 1, r)
````

下面这段代码是一个“**单指针、双边循环**”版本的快速排序（Quick Sort）实现。  
它把“**分区（partition）**”过程拆成了两步：

1. 先把 pivot 放到最左边 `nums[l]`。  
2. 用两条指针 `i`（从左向右）和 `j`（从右向左）把区间 `[l+1 … r]` 扫描一遍，最终把 pivot 插回到中间某个位置 `new_pivot`，使得  
   `nums[l … new_pivot-1] ≤ pivot_val ≤ nums[new_pivot+1 … r]`。

下面按行逐字拆开讲清每一句在干什么、为什么、以及潜在的小问题。

----------------
```python
def quickSort(nums: List[int], l: int, r: int) -> bool:
```
- 函数签名写返回 `bool`，但函数体里没有任何 `return True/False`，实际返回值永远是 `None`。  
- 更合理的签名应是 `-> None` 或干脆省略。

----------------
```python
    if l >= r:
        return
```
- 递归边界：区间长度 ≤1 时已经天然有序，直接结束。

----------------
```python
    pivot = random.randint(l, r)
    pivot_val = nums[pivot]
```
- 随机选一个基准值，避免最坏 O(n²) 发生。  
- `pivot` 是**下标**，`pivot_val` 才是真正的值。

----------------
```python
    nums[l], nums[pivot] = nums[pivot], nums[l]
```
- 把 pivot 元素交换到最左边，暂时腾出一个“空位”方便后面双边扫描。  
- 此时 `nums[l]` 就是 pivot 值。

----------------
```python
    i, j = l + 1, r
```
- 两条指针初始化：  
  - `i` 从左向右，指向“当前待检查”元素。  
  - `j` 从右向左，指向“当前待检查”元素。  
- 区间 `[l+1 … r]` 是我们真正要扫描的范围。

----------------
```python
    while True:
        while i < j and nums[j] >= pivot_val:
            j -= 1
        while i < j and nums[i] <= pivot_val:
            i += 1
        if i == j:
            break
        nums[i], nums[j] = nums[j], nums[i]
```
- **双边循环**核心：  
  1. 先让 `j` 向左走，直到找到一个小于 `pivot_val` 的元素（或 `i==j`）。  
  2. 再让 `i` 向右走，直到找到一个大于 `pivot_val` 的元素（或 `i==j`）。  
  3. 如果两条指针还没碰头，就把这两个“越位”元素交换，继续循环。  

- 循环结束后，`i==j` 的位置就是“分水岭”：  
  - 左侧 `[l+1 … i-1]` 全部 ≤ pivot_val  
  - 右侧 `[i+1 … r]` 全部 ≥ pivot_val  
  - 位置 `i` 本身可能属于任意一侧，需要再判断一次。

----------------
```python
    new_pivot = i if nums[i] <= nums[l] else i - 1
```
- 决定 pivot 最终应该插到哪里。  
- 因为 `nums[l]` 存的是 pivot 值，所以：  
  - 如果 `nums[i] ≤ pivot_val`，说明 `i` 这个位置可以归入“左半部分”，于是把 pivot 放到 `i`。  
  - 否则 `i` 位置元素大于 pivot，pivot 只能放到 `i-1`。  

- 这句写法等价于“把 `i` 向左回退到最后一个 ≤ pivot 的位置”。

----------------
```python
    nums[l], nums[new_pivot] = nums[new_pivot], nums[l]
```
- 把 pivot 从临时仓库 `l` 搬回真正的“中间”位置 `new_pivot`。  
- 搬完后满足：  
  `nums[l … new_pivot-1] ≤ pivot_val ≤ nums[new_pivot+1 … r]`

----------------
```python
    quickSort(nums, l, new_pivot - 1)
    quickSort(nums, new_pivot + 1, r)
```
- 对左右两个子区间递归调用快排。

----------------
**示例**

以 `nums = [6, 3, 9, 2, 7, 5]` 为例，假设随机 pivot 选到 `5`（下标 5）：

```
step 0: pivot_val = 5, 先把 5 换到最左  
        [5, 3, 9, 2, 7, 6]        l=0, r=5, i=1, j=5

step 1: j 向左找到 2（<5），停住；i 向右找到 9（>5），停住  
        [5, 3, 9, 2, 7, 6]        i=2, j=3

step 2: 交换 9 和 2  
        [5, 3, 2, 9, 7, 6]        i=2, j=3

step 3: 继续移动指针，i==j 退出循环  
        i=j=3, nums[3]=9 > 5 ⇒ new_pivot = i-1 = 2

step 4: 把 pivot 5 放回到 new_pivot=2  
        [2, 3, 5, 9, 7, 6]
        左子区间 [2,3]  右子区间 [9,7,6]
```

递归下去即可。

````python
"""
归并排序
"""
def mergeSort(nums: List[int], cache: List[int], l: int, r: int) -> bool:
    if l >= r:
        return
    mid = (l + r) // 2
    mergeSort(nums, cache, l, mid)
    mergeSort(nums, cache, mid + 1, r)

    i, j = l, mid + 1
    for pos in range(l, r + 1):
        if j > r or (i <= mid and nums[i] <= nums[j]):
            cache[pos] = nums[i]
            i += 1
        else: 
            cache[pos] = nums[j]
            j += 1
    nums[l:r+1] = cache[l:r+1]
````

# 归并排序（Merge Sort）—— 手把手详解

## 1. 算法思想
- **分治（Divide & Conquer）**  
  1. 分：把待排序区间 `[l, r]` 从中间劈成两半 `[l, mid]` 与 `[mid+1, r]`  
  2. 治：递归地对左右两半分别排序  
  3. 合：把两个**已经有序**的子区间合并成一个有序区间（merge 阶段）

- **稳定性**  
  合并时当 `nums[i] == nums[j]` 优先取左半部分元素，因而相等元素相对顺序不变，**稳定排序**

- **时间 / 空间复杂度**  
  - 时间：每次合并线性扫描，共 `log n` 层 ⇒ **O(n log n)**  
  - 空间：需要一段与原始数组等长的辅助缓存 ⇒ **O(n)**（非原地）

## 2. 代码逐行剖析
```python
from typing import List

def mergeSort(nums: List[int], cache: List[int], l: int, r: int) -> None:
    # 2.1 递归出口：区间长度为 0 或 1 时已经天然有序
    if l >= r:
        return

    # 2.2 分：找到中点（向下取整）
    mid = (l + r) // 2

    # 2.3 治：递归排序左右子区间
    mergeSort(nums, cache, l, mid)      # 左半 [l, mid]
    mergeSort(nums, cache, mid + 1, r)  # 右半 [mid+1, r]

    # 2.4 合：双指针合并到 cache[]，再写回原数组
    i, j = l, mid + 1          # i 扫描左半，j 扫描右半
    for pos in range(l, r + 1):
        # 右侧已取完 或 左侧当前元素更小/等于，则取左侧
        if j > r or (i <= mid and nums[i] <= nums[j]):
            cache[pos] = nums[i]
            i += 1
        else:                  # 否则取右侧
            cache[pos] = nums[j]
            j += 1

    # 2.5 把排好序的片段复制回原数组，保证后续上层合并拿到的是有序数据
    nums[l:r+1] = cache[l:r+1]
```

## 5. 常见疑问 Q&A
**Q1**: 能否省掉 `cache` 数组？  
**A**: 归并排序的合并阶段必须额外缓存，否则无法在 O(n) 时间内完成合并。若追求原地，可考虑 in-place 归并（复杂且常数大），或直接换用堆排序 / 快速排序。

**Q2**: 为什么 `nums[l:r+1] = cache[l:r+1]` 这一步不能少？  
**A**: 下一层合并依赖 `nums` 当前层已经有序。如果只写 `cache` 而不复制回去，上层再合并时拿到的 `nums` 仍是旧的无序数据。

**Q3**: 区间划分用 `mid = (l + r) // 2` 会溢出吗？  
**A**: 在 Python 中整数精度无限制，不会溢出；C/C++/Java 可改写为 `mid = l + (r - l) // 2` 防止 `(l + r)` 溢出。

````python
"""
215 Kth Largest Element in an Array
在一个未排序的数组中，找到第k大的数字
"""
# 输入一个数组和一个目标值k，输出第k大的数字。题目默认一定有解
def findKthLargest(nums: List[int], k: int) -> int:
    # random.choice() 从非空序列中随机返回一个元素
    pivot_val = random.choice(nums)
    # 初始化三个列表，pivot大于的，pivot等于的，pivot小于的
    larger, equal, smaller = [], [], []
    # 遍历列表所有元素，把元素添加到对应的列表里
    for num in nums:
        if num > pivot_val:
            larger.append(num)
        elif num < pivot_val:
            smaller.append(num)
        else:
            equal.append(num)

    # len(larger)代表pivot是第几大的元素
    # 如果k小于这个值，那么只需要在larger这个子序列里继续找就好
    if k <= len(larger):
        return findKthLargest(larger, k)
    # 如果k比len(larger)+len(equal)都大，说明k在smaller这个子序列里
    if k > len(larger) + len(equal):
        # 此时应该找子序列里第k-len(larger)-len(equal)的数字
        return findKthLargest(smaller, k - len(larger) - len(equal))
    # 如果不在larger里也不在smaller里，那么就在equal里，直接return即可
    return pivot_val
````

````python
from collections import Counter
````

````python
"""
347 Top K Frequent Elements
给定一个数组，求前k个最频繁的数字
"""
# 输入是一个数组和一个目标值k。输出是一个长度为k的数组。

def topKFrequent(nums: List[int], k: int) -> List[int]:
    # 统计元素的频次，映射为字典
    counts = Counter(nums)
    # 定义桶，为字典类型
    buckets = dict()

    # 对频次字典中的所有元素
    for num, count in counts.items():
        # 每个桶的key是以频次count作为编号的
        # 如果该频次的桶里出现过，则把对应数字加在桶里
        if count in buckets:
            buckets[count].append(num)
        # 如果没有出现该频次的桶，则初始化对应的桶，并把对应数字放在桶里
        else:
            buckets[count] = [num]
    # 初始化空列表
    top_k = []
    # 对所有频率（从0到len(nums）)逆序查找
    for count in range(len(nums), 0, -1):
        # 如果频率在buckets中出现
        if count in buckets:
            # 则top_k列表要加上该频率桶中所有的数字列表
            top_k += buckets[count]
            # 如果加上之后，发现频率最高数字的数量已经大于k
            if len(top_k) >= k:
                # 则把前k个数字返回
                return top_k[:k]
    return top_k[:k]
````
