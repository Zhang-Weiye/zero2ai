**表1-5  Python 集支持的运算**

| 运算名 | 运算符         | 解释                                         |
| ------ | -------------- | -------------------------------------------- |
| 成员   | in             | 询问集中是否有某元素                         |
| 长度   | len            | 获取集的元素个数                             |
| 并集   | \|             | 返回一个包含aset与otherset所有元素的新集      |
| 交集   | &              | 返回一个包含aset与otherset共有元素的新集      |
| 差集   | -              | 返回一个集，其中包含只出现在aset中的元素      |
| 子集   | <=             | 询问aset中的所有元素是否都在otherset中        |

**表1-6  Python 集提供的方法**

| 方法名        | 用法                          | 解释                                         |
| ------------- | ----------------------------- | -------------------------------------------- |
| union         | aset.union(otherset)          | 返回一个包含aset和otherset所有元素的集        |
| intersection  | aset.intersection(otherset)   | 返回一个仅包含两个集共有元素的集              |
| difference    | aset.difference(otherset)     | 返回一个集，其中仅包含只出现在aset中的元素    |
| issubset      | aset.issubset(otherset)       | 询问aset是否为otherset的子集                  |
| add           | aset.add(item)                | 向aset添加一个元素                            |
| remove        | aset.remove(item)             | 将item从aset中移除                            |
| pop           | aset.pop()                    | 随机移除aset中的一个元素                      |
| clear         | aset.clear()                  | 清除aset中的所有元素                          |

```python
## Python 集合运算演示

# 创建集合
mySet = {False, 4.5, 3, 6, 'cat'}
print("mySet:", mySet)

yourSet = {99, 3, 100}
print("yourSet:", yourSet)

# 并集
print("mySet.union(yourSet):", mySet.union(yourSet))
print("mySet | yourSet:", mySet | yourSet)

# 交集
print("mySet.intersection(yourSet):", mySet.intersection(yourSet))
print("mySet & yourSet:", mySet & yourSet)

# 差集
print("mySet.difference(yourSet):", mySet.difference(yourSet))
print("mySet - yourSet:", mySet - yourSet)

# 子集判断
print("{3,100}.issubset(yourSet):", {3,100}.issubset(yourSet))
print("{3,100} <= yourSet:", {3,100} <= yourSet)

# 添加元素
mySet.add("house")
print("mySet after add('house'):", mySet)

# 移除元素
mySet.remove(4.5)
print("mySet after remove(4.5):", mySet)

# 随机弹出一个元素
popped = mySet.pop()
print("popped element:", popped)
print("mySet after pop():", mySet)

# 清空集合
mySet.clear()
print("mySet after clear():", mySet)
```

**表1-7  Python 字典支持的运算**

| 运算名 | 运算符           | 解释                                         |
| ------ | ---------------- | -------------------------------------------- |
| 索引   | myDict[k]        | 返回与k相关联的值，如果没有则报错            |
| 成员   | key in adict     | 如果key在字典中，返回True，否则返回False     |
| 删除   | del adict[key]   | 删除字典中指定key对应的键值对                |

```python
phoneext = {'david': 1410, 'brad': 1137}
print("phoneext:", phoneext)

# 获取所有键
print("phoneext.keys():", phoneext.keys())
print("list(phoneext.keys()):", list(phoneext.keys()))

# 获取所有值
print("phoneext.values():", phoneext.values())
print("list(phoneext.values()):", list(phoneext.values()))

# 获取所有键值对
print("phoneext.items():", phoneext.items())
print("list(phoneext.items()):", list(phoneext.items()))

# 安全获取不存在的键
print('phoneext.get("kent"):', phoneext.get("kent"))
print('phoneext.get("kent", "NO ENTRY"):', phoneext.get("kent", "NO ENTRY"))
```

**表1-10  格式化修改符**

| 修改符   | 例子         | 解释                                         |
| -------- | ------------ | -------------------------------------------- |
| 数字     | %20d         | 将值放在20个字符宽的区域中                   |
| -        | %-20d        | 将值放在20个字符宽的区域中，并且左对齐       |
| +        | %+20d        | 将值放在20个字符宽的区域中，并且右对齐       |
| 0        | %020d        | 将值放在20个字符宽的区域中，并在前面补上0    |
| .        | %20.2f       | 将值放在20个字符宽的区域中，并且保留小数点后2位 |
| (name)   | %(name)d     | 从字典中获取name                             |

```python
price = 24
item = "banana"

print("The %s costs %d cents" % (item, price))
print("The %+10s costs %5.2f cents" % (item, price))
print("The %+10s costs %10.2f cents" % (item, price))

itemdict = {"item": "banana", "cost": 24}
print("The %(item)s costs %(cost)7.1f cents" % itemdict)
```
