```python
# 7 8 9
class Fraction():
    """
    添加方法radd, iadd, repr
    """
    def __init__(self, num, den):
        # if not isinstance(num, int) or not isinstance(den, int):
        #     raise ValueError
        if den == 0:
            raise ZeroDivisionError

        if den < 0:
            num, den = -num, -den
        factor = self._findMaxFactor(abs(num), abs(den))
        self.num = num //factor
        self.den = den //factor

    @staticmethod
    def _findMaxFactor(m, d):
    # 5, 2
    # 2-1, 2,1
    # 2-0, 1,0
        while d != 0:
            m, d = d, m%d
        return m

    def getNum(self):
        return self.num

    def getDen(self):
        return self.den

    def show(self):
        print(f"{self.num} / {self.den}")

    def __add__(self, frac: "Fraction") -> "Fraction":
        print("Now __add__ is running...")
        if isinstance(frac, Fraction):
            num = self.num*frac.den + frac.num*self.den
            den = self.den*frac.den
            return Fraction(num, den)
        else:
            return NotImplemented


    def __sub__(self, frac: "Fraction") -> "Fraction":
        num = self.num*frac.den - frac.num*self.den
        den = self.den*frac.den
        return Fraction(num, den)
    
    def __mul__(self, frac: "Fraction") -> "Fraction":
        num = self.num*frac.num
        den = self.den*frac.den
        return Fraction(num, den)
    
    def __truediv__(self, frac: "Fraction") -> "Fraction":
        num = self.num*frac.den
        den = self.den*frac.num
        return Fraction(num, den)

    def __lt__(self, frac: "Fraction") -> bool:
        # less than
        one_result = self.num * frac.den
        other_result = self.den * frac.num
        return one_result < other_result
    
    def __le__(self, frac: "Fraction") -> bool:
        # less than or equal
        one_result = self.num * frac.den
        other_result = self.den * frac.num
        return one_result <= other_result
    
    def __eq__(self, frac: "Fraction") -> bool:
        # equal
        one_result = self.num * frac.den
        other_result = self.den * frac.num
        return one_result == other_result
    
    def __ne__(self, frac: "Fraction") -> bool:
        # not equal
        one_result = self.num * frac.den
        other_result = self.den * frac.num
        return one_result != other_result
    
    def __gt__(self, frac: "Fraction") -> bool:
        # greater than
        one_result = self.num * frac.den
        other_result = self.den * frac.num
        return one_result > other_result
    
    def __ge__(self, frac: "Fraction") -> bool:
        # greater than or equal
        one_result = self.num * frac.den
        other_result = self.den * frac.num
        return one_result >= other_result
    
    def __radd__(self, frac) -> "Fraction":
        #  反向加法（应对 int/float + Fraction）
        print("Now __radd__ is running...")
        if isinstance(frac, Fraction):
            return self + frac
        elif isinstance(frac, (int, float)):
            return self + Fraction(frac, 1)
        else:
            raise ValueError
        
    def __iadd__(self, frac) -> "Fraction":
        # +=
        print("Now __iadd_ is running...")
        if isinstance(frac, Fraction):
            result = self + frac
            self.num, self.den = result.num, result.den
            return self
        elif isinstance(frac, (int, float)):
            result = self + Fraction(frac, 1)
            self.num, self.den = result.num, result.den
            return self
        else:
            raise ValueError
        
    def __repr__(self) -> str:
        # 交互式解释器友好输出
        return f"Fraction({self.num}, {self.den})"
    
frac1 = Fraction(1, 2)
frac2 = 1  + frac1 
frac3 = 0.5 + frac1
frac3 += frac1
frac3
        
```
